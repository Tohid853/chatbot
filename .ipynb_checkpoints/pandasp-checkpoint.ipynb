```python
import json
import random
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
import joblib
with open("d.json","r") as file:
     data=json.load(file)
sentences=[]
labels=[]
responses={}
for i in data["intents"]:
    tag=i["tag"]
    responses[tag]=i["responses"]
    for j in i['patterns']:
        sentences.append(j)
        labels.append(tag)
vector=TfidfVectorizer()
x=vector.fit_transform(sentences)
model=MultinomialNB()
model.fit(x,labels)
joblib.dump(model,"chatbot.pkl")
joblib.dump(vector,"vector.pkl")
joblib.dump(responses,"res.pkl")
    
```
